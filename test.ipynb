```python
import matplotlib
matplotlib.use("WebAgg")
import pandas as pd
import matplotlib.pyplot as plt
from matplotlib.animation import FuncAnimation
import matplotlib.pyplot as plt
```

```python
# instellen van een webserver waar matplotlib de grafieken kan weergeven
plt.rcParams["webagg.address"] = "0.0.0.0"
plt.rcParams["webagg.port"] = 8988
plt.rcParams["webagg.open_in_browser"] = False
```

```python
# CSV-bestand inlezen
df = pd.read_csv("bombbba.mqa")
```

```python
# Posities van de drie punten
x1, y1 = df["Point #1.X"], df["Point #1.Y"]
x2, y2 = df["Point #2.X"], df["Point #2.Y"]
x3, y3 = df["Point #3.X"], df["Point #3.Y"]

# Relatieve posities ten opzichte van bal 2
x1_relative = df["Point #1.X"] - df["Point #2.X"]
y1_relative = df["Point #1.Y"] - df["Point #2.Y"]

x2_relative = 0
y2_relative = 0

x3_relative = df["Point #3.X"] - df["Point #2.X"]
y3_relative = df["Point #3.Y"] - df["Point #2.Y"]

time = df["Time"]
```

```python
# Hoeken
```

```python
# Figuur aanmaken
fig, ax = plt.subplots()

ax.set_xlim(df[["Point #1.X", "Point #2.X", "Point #3.X"]].min().min(),
            df[["Point #1.X", "Point #2.X", "Point #3.X"]].max().max())

ax.set_ylim(df[["Point #1.Y", "Point #2.Y", "Point #3.Y"]].min().min(),
            df[["Point #1.Y", "Point #2.Y", "Point #3.Y"]].max().max())

ax.set_aspect("equal")
ax.set_xlabel("X")
ax.set_ylabel("Y")

# Drie bewegende punten
point1, = ax.plot([], [], "ro", label="Punt 1")
point2, = ax.plot([], [], "bo", label="Punt 2")
point3, = ax.plot([], [], "go", label="Punt 3")

ax.legend()

# Animatiefunctie
def update(frame):
    point1.set_data([x1.iloc[frame]], [y1.iloc[frame]])
    point2.set_data([x2.iloc[frame]], [y2.iloc[frame]])
    point3.set_data([x3.iloc[frame]], [y3.iloc[frame]])

    ax.set_title(f"Tijd: {time.iloc[frame]:.2f} s")

    return point1, point2, point3

# Animatie maken
ani = FuncAnimation(
    fig,
    update,
    frames=len(df),
    interval=50,
    blit=False
)

plt.show()
```

```python
# Tijd
time = df["Time"]


# Figuur aanmaken
fig, ax = plt.subplots()

# Grenzen bepalen op basis van relatieve posities
xmin = min(x1.min(), x3.min(), 0)
xmax = max(x1.max(), x3.max(), 0)
ymin = min(y1.min(), y3.min(), 0)
ymax = max(y1.max(), y3.max(), 0)

# Kleine marge toevoegen
margin = 0.1 * max(xmax - xmin, ymax - ymin, 1)

ax.set_xlim(xmin - margin, xmax + margin)
ax.set_ylim(ymin - margin, ymax + margin)

ax.set_aspect("equal")
ax.set_xlabel("Relatieve X")
ax.set_ylabel("Relatieve Y")
ax.grid(True, alpha=0.3)

# Drie bewegende punten
point1, = ax.plot([], [], "ro", label="Bal 1", markersize=8)
point2, = ax.plot([], [], "bo", label="Bal 2 (oorsprong)", markersize=8)
point3, = ax.plot([], [], "go", label="Bal 3", markersize=8)

# Verbindingslijnen
line12, = ax.plot([], [], "r-", linewidth=1.5)
line23, = ax.plot([], [], "g-", linewidth=1.5)

ax.legend()

# Animatiefunctie
def update(frame):

    # Posities van de ballen
    x1_now = x1.iloc[frame]
    y1_now = y1.iloc[frame]

    x2_now = 0
    y2_now = 0

    x3_now = x3.iloc[frame]
    y3_now = y3.iloc[frame]

    # Punten bijwerken
    point1.set_data([x1_now], [y1_now])
    point2.set_data([x2_now], [y2_now])
    point3.set_data([x3_now], [y3_now])

    # Verbindingslijnen bijwerken
    line12.set_data(
        [x1_now, x2_now],
        [y1_now, y2_now]
    )

    line23.set_data(
        [x2_now, x3_now],
        [y2_now, y3_now]
    )

    # Tijd weergeven
    ax.set_title(f"Tijd: {time.iloc[frame]:.2f} s")

    return point1, point2, point3, line12, line23

# Animatie maken
ani = FuncAnimation(
    fig,
    update,
    frames=len(df),
    interval=50,
    blit=False,
    repeat=True
)

plt.show()
```
